```python
import ipywidgets
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import math
```

```python
x = np.random.uniform(0, 5, size = 100)
noise = np.random.normal(size = 100)
```

```python
def plot_fct(w=1):
  y = np.long(x) + w * noise
  plt.scatter (x,y)
plot_fct()
```

```python
from sklearn.datasets import make_moons
def plot_moons(n_samples=200, noise=0):
  moons=make_moons(n_samples, noise=noise)
  x, y =moons[0], moons[1]
  plt.scatter(x[:,0], x[:,1], c=y)
```

```python
ipywidgets.interact(plot_moons , samples =[50,200,500,1000],noise=(0,2,0.25))
```
