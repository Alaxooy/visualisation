```python
!pip install dash yfinance
from dash import Dash
import pandas as pd
from dash import dcc, html, Input, Output
import yfinance as yf
import plotly.express as px
```

```python
# Initialize the app
app = Dash(__name__)
```

```python
app.layout = html.Div([
    html.H1("Stock Price Analysis Dashboard"),

    html.Label("Select Stock Symbol:"),
    dcc.Dropdown(
        id='stock-selector',
        options=[
            {'label': 'AAPL', 'value': 'AAPL'},
            {'label': 'GOOGLE', 'value': 'GOOG'},
            {'label': 'MSFT', 'value': 'MSFT'},
            {'label': 'NETFLIX', 'value': 'NFLX'},
        ],
        value='AAPL'
    ),

    dcc.Graph(id='stock-price-chart'),
])
```

```python
@app.callback(
    Output('stock-price-chart', 'figure'),
    [Input('stock-selector', 'value')]
)
def update_chart(selected_stock):
    try:

        # Online Dataset
        # stock_data = yf.download(selected_stock, start='2022-06-01', end='2024-06-01')

        # Local Dataset
        stock_data = pd.read_csv('stocks.csv', parse_dates=['Date'])
        stock_data = stock_data[stock_data['Ticker'] == selected_stock]

        # Optional: filter by date range
        stock_data = stock_data[(stock_data['Date'] >= '2022-06-01') & (stock_data['Date'] <= '2024-06-01')]

        if stock_data.empty:
            fig = {
                "data": [],
                "layout": {
                    "title": f"No data available for {selected_stock}"
                }
            }
        else:
            fig = px.line(stock_data, x='Date', y='Close',
                          title=f'{selected_stock} Stock Price Analysis')
    except Exception as e:
        fig = {
            "data": [],
            "layout": {
                "title": f"Error loading data: {e}"
            }
        }

    return fig
```

```python
if __name__ == '__main__':
  app.run(port=8050)
```
