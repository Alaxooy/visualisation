```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
titanic = sns.load_dataset('titanic')
```

```python
print(titanic.head())
```

```python
sns.lineplot(x='age', y='survived' , data=titanic)
plt.title('age vs survival ')
plt.xlabel('age')
plt.ylabel('survival')
plt.show()
```

```python
sns.histplot(x='age', data=titanic , hue='survived', bins= 10, kde= True)
plt.title('age distribution')
plt.xlabel('age')
plt.ylabel('count')
plt.show()
```
